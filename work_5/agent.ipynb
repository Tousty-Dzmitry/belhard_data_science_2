```python
import numpy as np
import random
import matplotlib.pyplot as plt
import matplotlib.patches as patches

# Класс для создания игровой среды (сетки)
class Environment:
    def __init__(self, size=5):
        # Инициализация среды с заданным размером
        self.size = size
        # Создание сетки (матрицы) заполненной нулями
        self.grid = np.zeros((size, size))
        # Размещение сокровищ на сетке
        self._place_treasures()
        # Размещение ловушек на сетке
        self._place_traps()

    def _place_treasures(self):
        """Размещает 3 сокровища (обозначаются 1) в случайных позициях сетки"""
        for _ in range(3):
            # Генерация случайных координат
            x, y = random.randint(0, self.size - 1), random.randint(0, self.size - 1)
            # Установка значения 1 (сокровище) в выбранной позиции
            self.grid[x][y] = 1

    def _place_traps(self):
        """Размещает 3 ловушки (обозначаются -1) в случайных пустых позициях"""
        for _ in range(3):
            while True:
                # Генерация случайных координат
                x, y = random.randint(0, self.size - 1), random.randint(0, self.size - 1)
                # Проверка, что клетка пустая (0)
                if self.grid[x][y] == 0:
                    # Установка ловушки (-1)
                    self.grid[x][y] = -1
                    break  # Выход из цикла после размещения

    def get_state(self, pos):
        """Возвращает состояние клетки по координатам (x,y)"""
        x, y = pos
        # Проверка, что координаты в пределах сетки
        if 0 <= x < self.size and 0 <= y < self.size:
            return self.grid[x][y]  # Возврат значения клетки
        return -2  # Возврат -2 для клеток за пределами сетки (границы)

    def render(self, agent):
        """Визуализирует текущее состояние среды с агентом"""
        # Создание фигуры и осей для отрисовки
        fig, ax = plt.subplots()
        # Установка границ отрисовки
        ax.set_xlim(0, self.size)
        ax.set_ylim(0, self.size)
        # Настройка сетки
        ax.set_xticks(range(self.size + 1))
        ax.set_yticks(range(self.size + 1))
        ax.grid(True)
        ax.set_aspect('equal')
        ax.invert_yaxis()  # Инверсия оси Y (чтобы (0,0) был вверху слева)

        # Отрисовка каждой клетки сетки
        for x in range(self.size):
            for y in range(self.size):
                value = self.grid[x][y]  # Значение клетки
                pos = (x, y)  # Позиция клетки

                # Выбор цвета в зависимости от типа клетки
                if value == 1:  # Сокровище
                    color = 'gold'
                elif value == -1:  # Ловушка
                    color = 'red'
                elif pos in agent.visited:  # Посещенная клетка
                    color = 'lightgray'
                else:  # Неизведанная клетка
                    color = 'white'

                # Создание прямоугольника для клетки
                rect = patches.Rectangle((y, x), 1, 1, facecolor=color, edgecolor='black')
                ax.add_patch(rect)  # Добавление прямоугольника на график

        # Отрисовка позиции агента (синий кружок)
        ax.plot(agent.position[1] + 0.5, agent.position[0] + 0.5, 'o', color='blue', markersize=12)
        # Установка заголовка с информацией о состоянии агента
        ax.set_title(f'Позиция: {agent.position} | Счёт: {agent.score} | Штраф: {agent.penalty} | {"(!)" if not agent.alive else ""}')
        plt.pause(0.5)  # Пауза для анимации
        plt.clf()  # Очистка фигуры для следующего кадра

# Класс ИИ-агента
class SimpleAIAgent:
    def __init__(self, env):
        # Инициализация агента с привязкой к среде
        self.env = env
        self.position = (0, 0)  # Начальная позиция
        self.memory = {}  # Память о посещенных местах
        self.visited = set()  # Множество посещенных клеток
        self.score = 0  # Счет (найденные сокровища)
        self.alive = True  # Флаг жизни агента
        self.penalty = 0  # Штрафные очки

    def perceive(self):
        """Сбор информации об окружающих клетках"""
        x, y = self.position
        # Создание словаря с состоянием соседних клеток
        surroundings = {
            'current': self.env.get_state((x, y)),  # Текущая клетка
            'north': self.env.get_state((x - 1, y)),  # Север
            'south': self.env.get_state((x + 1, y)),  # Юг
            'west': self.env.get_state((x, y - 1)),  # Запад
            'east': self.env.get_state((x, y + 1))  # Восток
        }
        # Сохранение информации в памяти
        self.memory[self.position] = surroundings
        # Добавление текущей позиции в посещенные
        self.visited.add(self.position)
        return surroundings

    def plan_action(self):
        """Планирование следующего действия на основе текущего состояния"""
        x, y = self.position
        # Если на текущей клетке сокровище - собираем
        if self.env.get_state((x, y)) == 1:
            return 'collect'

        # Сбор возможных ходов
        possible_moves = []
        # Варианты движений с соответствующими координатами
        moves = [('вверх', (x - 1, y)),
                 ('вниз', (x + 1, y)),
                 ('влево', (x, y - 1)),
                 ('вправо', (x, y + 1))]

        for move, pos in moves:
            state = self.env.get_state(pos)
            # Проверка что клетка не посещена и не является ловушкой/границей
            if pos not in self.visited and state >= 0:
                possible_moves.append((move, pos))

        # Если есть доступные ходы - случайный выбор
        if possible_moves:
            return random.choice(possible_moves)[0]

        # Если нет доступных ходов - случайное движение
        return random.choice(['вверх', 'вниз', 'влево', 'вправо'])

    def execute_action(self, action):
        """Выполнение запланированного действия"""
        x, y = self.position
        # Действие сбора сокровища
        if action == 'collect' and self.env.get_state((x, y)) == 1:
            self.env.grid[x][y] = 0  # Удаление сокровища
            self.score += 1  # Увеличение счета
            return f"Сокровище найдено! Всего: {self.score}"

        # Определение новой позиции после движения
        new_position = {
            'вверх': (x - 1, y),
            'вниз': (x + 1, y),
            'влево': (x, y - 1),
            'вправо': (x, y + 1)
        }.get(action, (x, y))  # По умолчанию - остаться на месте

        # Проверка состояния новой позиции
        state = self.env.get_state(new_position)

        # Обработка разных состояний клетки:
        if state == -2:  # Граница карты
            self.penalty += 1
            return "Столкновение с границей!"
        elif state == -1:  # Ловушка
            self.penalty += 5
            self.position = new_position
            self.alive = False  # Агент погиб
            return "Попал в ловушку! Агент уничтожен!"
        elif state == 0:  # Пустая клетка
            self.penalty += 1  # Штраф за перемещение

        # Обновление позиции агента
        self.position = new_position
        return f"Перемещение: {action.upper()}"

    def learn(self):
        """Обновление знаний агента на основе опыта"""
        for pos, data in self.memory.items():
            # Помечаем клетки без сокровищ как посещенные
            if data['current'] <= 0:
                self.visited.add(pos)

    def run(self, max_steps=20):
        """Основной цикл работы агента"""
        print("Агент начал исследование!")
        # Цикл по шагам эпизода
        for step in range(max_steps):
            # Проверка жив ли агент
            if not self.alive:
                print("Агент погиб.")
                break

            print(f"\nШаг {step + 1}: Позиция {self.position}")
            # 1. Восприятие окружающей среды
            self.perceive()
            # 2. Планирование действия
            action = self.plan_action()
            # 3. Выполнение действия
            result = self.execute_action(action)
            print(f"Действие: {action} | Результат: {result}")
            # 4. Визуализация состояния
            self.env.render(self)
            # 5. Обновление знаний
            self.learn()

            # Проверка условия завершения (все сокровища найдены)
            if self.score >= 3:
                print("Все сокровища найдены!")
                break

        # Вывод итоговой карты знаний
        print("\nИтоговая карта знаний:")
        for x in range(self.env.size):
            for y in range(self.env.size):
                pos = (x, y)
                # Символьное представление клеток:
                if pos == self.position:  # Агент
                    cell = "A"
                elif self.env.grid[x][y] == 1:  # Сокровище
                    cell = "T"
                elif self.env.grid[x][y] == -1:  # Ловушка
                    cell = "X"
                elif pos in self.visited:  # Посещенная клетка
                    cell = "V"
                else:  # Неизвестная клетка
                    cell = "?"
                print(cell, end=" ")
            print()  # Новая строка

# Основной блок выполнения
plt.ion()  # Включение интерактивного режима matplotlib

# Настройки экспериментов
NUM_EPISODES = 5  # Количество эпизодов
results = []       # Результаты по эпизодам
penalties = []     # Штрафы по эпизодам

# Цикл по эпизодам
for episode in range(1, NUM_EPISODES + 1):
    print(f"\n=== ЭПИЗОД {episode} ===")
    # Создание новой среды и агента
    env = Environment(size=5)
    agent = SimpleAIAgent(env)
    # Запуск эпизода
    agent.run(max_steps=20)

    # Сохранение результатов эпизода
    results.append({
        "score": agent.score,        # Количество найденных сокровищ
        "alive": agent.alive,        # Выжил ли агент
        "steps": len(agent.visited)  # Количество посещенных клеток
    })
    penalties.append(agent.penalty)  # Штрафные очки

# Вывод итоговых результатов
print("\n=== РЕЗУЛЬТАТЫ ЭПИЗОДОВ ===")
for i, res in enumerate(results, 1):
    status = 'Выжил' if res['alive'] else 'Погиб'
    print(f"Эпизод {i}: Сокровища: {res['score']} | Шагов: {res['steps']} | Штраф: {penalties[i-1]} | {status}")

# Визуализация результатов экспериментов
scores = [res["score"] for res in results]  # Список сокровищ по эпизодам

plt.figure(figsize=(12, 5))  # Создание фигуры

# График найденных сокровищ
plt.subplot(1, 2, 1)  # Левый график
plt.bar(range(1, NUM_EPISODES + 1), scores, color='green')
plt.xlabel("Эпизод")
plt.ylabel("Найдено сокровищ")
plt.title("Количество сокровищ по эпизодам")
plt.ylim(0, 3)  # Ограничение по оси Y (0-3 сокровища)
plt.grid(True)   # Включение сетки

# График штрафных очков
plt.subplot(1, 2, 2)  # Правый график
plt.bar(range(1, NUM_EPISODES + 1), penalties, color='darkred')
plt.xlabel("Эпизод")
plt.ylabel("Штрафные очки")
plt.title("Штрафные очки по эпизодам")
plt.grid(True)  # Включение сетки

plt.tight_layout()  # Автоматическая настройка отступов
plt.show()  # Показ графиков
```
